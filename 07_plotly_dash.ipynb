# 07 — Plotly Dash

```python
import pandas as pd, plotly.express as px
df=pd.read_csv("../data/cleaned_spacex_falcon9.csv")
site="CCAFS SLC 40"; d=df[df.LaunchSite==site]
px.pie(d,names=d.Class.map({0:"Failure",1:"Success"}),title="Landing outcome").show()
px.scatter(d,x="PayloadMass",y="Class",hover_data=["FlightNumber","Orbit"],title="Payload mass vs landing outcome").show()
```

The interactive application is implemented in `code/app.py` with a launch-site dropdown, pie chart, and payload/outcome scatter.
