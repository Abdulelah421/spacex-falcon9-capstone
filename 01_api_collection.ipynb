# 01 — SpaceX API Data Collection
Falcon 9 first-stage landing prediction capstone.

```python
import requests, pandas as pd
BASE = "https://api.spacexdata.com/v4"
launches = requests.get(f"{BASE}/launches/past", timeout=30).json()
launches_df = pd.json_normalize(launches)
display(launches_df.head())
```

Workflow: GET historical launches → normalize JSON → join rocket/payload/launchpad metadata → filter Falcon 9 → export structured launch records. Packaged benchmark contains 90 Falcon 9 records.
