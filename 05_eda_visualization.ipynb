# 05 — EDA & Visualization

```python
import pandas as pd, matplotlib.pyplot as plt
df=pd.read_csv("../data/cleaned_spacex_falcon9.csv")
yearly=df.groupby("Year")["Class"].mean().mul(100)
yearly.plot(marker="o",title="Yearly Falcon 9 landing success rate")
plt.ylabel("Success rate (%)"); plt.tight_layout(); plt.show()
```

```python
site=df.groupby("LaunchSite").agg(launches=("Class","size"),success_rate=("Class","mean"))
site["success_rate"]*=100
display(site.sort_values("launches",ascending=False))
```

Insights: landing success improved strongly over time; payload is not a simple monotonic predictor because mission type, site, booster history, and era interact.
