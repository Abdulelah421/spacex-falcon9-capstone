# 04 — EDA with SQL

```python
import sqlite3, pandas as pd
df=pd.read_csv("../data/cleaned_spacex_falcon9.csv")
con=sqlite3.connect("../data/spacex.sqlite")
df.to_sql("SPACEXTBL",con,if_exists="replace",index=False)
queries={
"Launch counts by site":"SELECT LaunchSite, COUNT(*) AS launches FROM SPACEXTBL GROUP BY LaunchSite ORDER BY launches DESC",
"Success rate by site":"SELECT LaunchSite, ROUND(AVG(Class)*100,1) AS success_rate_pct FROM SPACEXTBL GROUP BY LaunchSite ORDER BY success_rate_pct DESC",
"Yearly performance":"SELECT Year, COUNT(*) AS launches, ROUND(AVG(Class)*100,1) AS success_rate_pct FROM SPACEXTBL GROUP BY Year ORDER BY Year",
"Orbit performance":"SELECT Orbit, COUNT(*) AS launches, ROUND(AVG(Class)*100,1) AS success_rate_pct FROM SPACEXTBL GROUP BY Orbit ORDER BY success_rate_pct DESC"}
for title,q in queries.items():
    print("\n"+title); display(pd.read_sql_query(q,con))
con.close()
```

Key results: KSC LC 39A = 22 launches / 77.3% success; VAFB SLC 4E = 13 / 76.9%; GTO = 27 / 51.9%.
